```python
import pickle
```

```python
loaded_model = pickle.load(open("finalized_model_RandomForest.sav",'rb'))
```

```python
Age_input = float(input("Age="))
al_input = float(input("Albumin="))
bgr_input = int(input("Blood Glucose="))
bu_input = int(input("Blood Urea="))
sc_input = float(input("Serum Creatinine="))
hemo_input = float(input("Haemoglobin="))
pcv_input = int(input("Packed cell Volume="))
wc_input = int(input("White blood cell count="))
```

```python
result = loaded_model.predict([[Age_input, al_input, bgr_input, bu_input, sc_input, hemo_input, pcv_input, wc_input]])
```
