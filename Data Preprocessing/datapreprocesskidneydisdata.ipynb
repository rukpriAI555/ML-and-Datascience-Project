```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        #print(columnName)
        if(dataset[columnName].dtype=='O'):
            #print("qual")
            qual.append(columnName)
        else:
            #print("quan")
            quan.append(columnName)
    return quan,qual
```

```python
quan, qual = quanQual(dataset)  # Now 'quan' and 'qual' are defined
```

```python
dataset[qual]
```

```python
dataset[quan]
```

```python
# Filling missing values for quan
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values = np.nan, strategy = 'mean')
imp.fit(dataset[quan])
df = imp.transform(dataset[quan])
df = pd.DataFrame(df, columns = quan)
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Filling missing values for qual
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values = np.nan, strategy = 'most_frequent')
imp.fit(dataset[qual])
df_qual = imp.transform(dataset[qual])
df_qual = pd.DataFrame(df_qual, columns = qual)
```

```python
df_qual
```

```python
df_qual.isnull().sum()
```

```python
two = [df, df_qual]
```

```python
no_null_dataset = pd.concat(two, axis=1)
```

```python
no_null_dataset.isnull().sum()
```

```python
print(no_null_dataset.columns.tolist())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_columns = [
    'rbc', 'pc', 'pcc', 'ba', 'htn',
    'dm', 'cad', 'appet', 'pe', 'ane'
]

for column in categorical_columns:
    no_null_dataset[column] = le.fit_transform(
    no_null_dataset[column].astype(str).str.strip()
    )

# Encode the target column separately
    no_null_dataset['classification'] = le.fit_transform(
    no_null_dataset['classification'].astype(str).str.strip()
)
```
